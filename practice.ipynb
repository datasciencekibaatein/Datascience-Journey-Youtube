### We use 'with'

read() function

```python
with open("student_data.txt",'r') as file:
    print(file.read())
```

readline()

```python
with open('student_data.txt','r') as file:
    first_line = file.readline()
    second_line = file.readline()

    print(first_line)
    print(second_line)
```

readlines()

```python
with open('student_data.txt','r') as file:
    lines = file.readlines()
    print(lines)
    print(type(lines))
```

## How w (write mode) is dangerous !

```python
with open('student_data.txt','w') as file:
    file.write("Neha,22,34000")
```

writelines()

```python
list = [
    "rahul,24,30000\n",
    'Krish,23,23000\n',
    'Neha,23,34000\n'
]

with open('student_data.txt','w') as file:
    file.writelines(list)
```

## a (append) mode

```python
with open("student_data.txt",'a') as file:
    file.write("Krishna,23,25000\n")
```

```python
list = [
    "Amit,22,40000\n",
    'Aaditya,24,22000\n',
    'Abhi,23,35000\n'
]

with open('student_data.txt','a') as file:
    file.writelines(list)
```

```python
with open("student_data.txt",'r') as file:
    print(file.read())
```

## CSV

```python
import csv
```

when you are using writerow, u put the single row in file

```python
with open("employee_data.csv",'w',newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name","Age","Salary"])
    writer.writerow(["Mohit",23,45000])
    writer.writerow(['Sonu',34,50000])
```

```python
with open("employee_data.csv",'w',newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Krishna",23,45000])
    writer.writerow(['Monu',34,50000])
```

writerows() , when we use this we can write multiple rows simultaneously

```python
data = [
    ["Krishna",23,45000],
    ['Monu',34,50000]
]

with open("employee_data.csv",'a',newline="") as file:
    writer = csv.writer(file)

    writer.writerows(data)
```

read the csv file

```python
with open("employee_data.csv",'r',newline="") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
with open("employee_data.csv",'r',newline="") as file:
    reader = csv.reader(file)
    
    next(reader)#skip the header
    
    for row in reader:
        print(row)
```

## JSON

```python
import json
```

dumps() returns the json string

```python
employee = {
    "name":"Sohit",
    "age":25
}

json_string=json.dumps(employee,indent=4)

print(json_string)
print(type(json_string))
```

dump() function write object of python in json file 

```python
import json

employee = {
    "name":"Ram",
    "Age":34,
    "Salary":34000
}


with open("employee.json",'w') as file:
    json.dump(
        employee,
        file,
        indent=4
    )
```

```python
with open("employee.json",'r' ) as f:
    data = json.load(f)

    print(data)
    print(type(data))
```

```python
import json

json_string = '''

{
    "name":"rahul",
    "age":34
}
'''

data = json.loads(json_string)
print(data)
```
