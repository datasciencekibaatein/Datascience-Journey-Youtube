````python
from student_operations import add_student, view_all_students, search_student, update_student, delete_student, create_student
from statistics_generator import class_statistics, subject_wise_analysis, get_top_performers
from file_manager import save_to_csv, load_from_csv, export_to_json

print("Modules imported.")
````

## Load existing data (if any)

````python
students = load_from_csv()
print(f"Students in system: {len(students)}")
````

## Add students — interactive

````python
add_student(students)
````

## Add sample students programmatically (for demo)

````python
sample_data = [
    ("R001", "Rahul Sharma",  "12A", {"Physics": 85, "Chemistry": 78, "Maths": 92, "English": 74, "Computer": 88}),
    ("R002", "Priya Verma",   "12A", {"Physics": 90, "Chemistry": 88, "Maths": 85, "English": 82, "Computer": 91}),
    ("R003", "Amit Kumar",    "12A", {"Physics": 65, "Chemistry": 70, "Maths": 68, "English": 72, "Computer": 75}),
    ("R004", "Sneha Patel",   "12A", {"Physics": 78, "Chemistry": 82, "Maths": 88, "English": 79, "Computer": 85}),
    ("R005", "Rohit Singh",   "12A", {"Physics": 45, "Chemistry": 50, "Maths": 30, "English": 55, "Computer": 60}),
    ("R006", "Kavya Reddy",   "12A", {"Physics": 95, "Chemistry": 92, "Maths": 98, "English": 89, "Computer": 94}),
]

for roll_no, name, cls, marks in sample_data:
    if not any(s['roll_no'] == roll_no for s in students):
        student = create_student(roll_no, name, cls, marks)
        students.append(student)
        print(f"Added: {student['name']} — {student['percentage']}% ({student['grade']})")
````

## View all students

````python
view_all_students(students)
````

## Search a student

````python
search_student(students)
````

## Class statistics

````python
class_statistics(students)
````

## Subject-wise analysis

````python
subject_wise_analysis(students)
````

## Top performers

````python
for i, s in enumerate(get_top_performers(students, n=3), start=1):
    print(f"{i}. {s['name']:<20} {s['percentage']}% ({s['grade']})")
````

## Update student marks

````python
update_student(students)
````

## Delete a student

````python
delete_student(students)
````

## Save and export

````python
save_to_csv(students)
export_to_json(students)
````
