# Exception Handling

Exception handling is mechanism that allows a program to handle runtime errors gracefully instead crashing.

## What is exception handling ?

An exception is an error that occurs during program execution.

```python
print("before program")
num = 5

res = num / 0

print(res)

print("program ended")
```

### why exception handling ?

Without exception handling

- Program crashes
- Remaining code doesn't execute
- Poor experience

|Exception|When it occurs ?|
|---|---|
|ValueError|Wrong value|
|TypeError|Wrong type operation|
|FileNotFoundError|File missing|
|ZeroDivisionError|Divide by zero|
|IndexError|List index out of range|
|KeyError|Dictionary key missing|
|NameError|Variable not found|
|AttributeError|Wrong Method|

#### ValueError

```python
try:
    num = int(input("Enter the number: "))
    print(num)
except ValueError as e:
    print(e)
    print("This text is from except block")

print("program ended")
```

#### TypeError

```python
try:
    result = "rahul"+23

except TypeError as e:
    print(e)

print("program ended")
```

#### FileNotFoundError

```python
try:
    with open("dummy.txt",'r') as file:
        data =file.read()
except FileNotFoundError as e:
    print(e)

    
```

#### ZeroDivisionError

```python
try:
    num = 5
    res = num / 0
    print(res)
except ZeroDivisionError as e:
    print(e)

print("program ended")
```

#### IndexError

```python
fruits = ["apple","guava","kivi",'melon','mango']

try:
    index = int(input("Enter the fruit list index: "))
    print(fruits[index])
except IndexError as e:
    print(e)
```

#### KeyError

```python
student = {
    "Name":"Rahul",
    "age":23,
    "gender":"male"
}

try:
    print(student["marks"])
except KeyError as e:
    print(e)
```

## Multiple Except block

```python
try:
    num = int(input("Enter the num:"))
    result = 100/num
    print(result)
except ValueError as e:
    print(e)
    print("i am in value error")
except ZeroDivisionError as e:
    print(e)
    print("zero division error")
```

```python
try:
    num = int(input("Enter the num:"))
    result = 100/num
    print(result)

except ZeroDivisionError as e:
    print(e)
    print("zero division error")

except Exception as e:
    print(e)
    print("i am in exception block")
```

### Else block

else block will be executed when there is no error in vulnerable block.

```python
try:
    num = int(input("Enter the num:"))
    result = 100/num
    print(result)

except ZeroDivisionError as e:
    print(e)
    print("zero division error")

except Exception as e:
    print(e)
    print("i am in exception block")

else:
    print("i run when there is no error")
```

### Finally block

```python
try:
    num = int(input("Enter the num:"))
    result = 100/num
    print(result)

except ZeroDivisionError as e:
    print(e)
    print("zero division error")

except Exception as e:
    print(e)
    print("i am in exception block")

finally:
    print("i always run")
```
