# filter with multiple conditions

Salary > 50000 and the city should be delhi

```python
df1[(df1['Salary']>50000) & (df1['City']=='Delhi') ]
```

```python
df1[(df1['Salary']>50000) | (df1['City']=='Delhi')]
```

```python
df1[~(df1['City']=='Delhi') ]
```
