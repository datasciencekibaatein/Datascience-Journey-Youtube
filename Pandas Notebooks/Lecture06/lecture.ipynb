# Merge

```python
orders = pd.read_excel("dataset.xlsx",sheet_name='Orders_Fact')
products = pd.read_excel("dataset.xlsx",sheet_name='Products')
customers = pd.read_excel("dataset.xlsx",sheet_name='Customers')

```

```python
orders
```

```python
products
```

```python
df = pd.merge(orders,customers,on='customer_id')
```

```python
products
```

```python
df = pd.merge(df,products,left_on="product",right_on="product_name")
```

```python
df['Sales'] = df['quantity'] * df['price']
```

```python
df
```

```python
df.groupby('category')['Sales'].sum()
```

## inner , left , right and outer

```python
orders
```

```python
customers
```

By default we use inner join

```python
pd.merge(orders,customers,on='customer_id')
```

```python
orders
```

left join, fetch all the records from left table, place nan in right if the data does not exist

```python
pd.merge(orders,customers,on='customer_id',how='left')
```

```python
customers
```

```python
pd.merge(orders,customers,on='customer_id',how='inner',validate="m:1")
```

Outer join

```python
pd.merge(orders,customers,on='customer_id',how='outer')
```
