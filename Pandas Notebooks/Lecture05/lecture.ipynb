## Group by function

mean,sum,count,max,min

```python
df.groupby('Gender')['Salary'].mean()
```

Salary Pool by department

```python
df.groupby('Department',as_index=False)['Salary'].sum()
```

## Group by with multiple aggregation

```python
df.groupby('City')['Salary'].agg(['sum','mean','min','max'])
```

## Grouping multiple columns

```python
temp = df.groupby(['City','Gender','Employment_Type'],as_index=False)['Salary'].mean()
```

```python
temp.head(30)
```

```python
df.groupby(['Gender','City'])[['Salary','Age']].sum()
```

## Custom function with group by

```python
df.groupby('Department')['Salary'].apply(lambda x:x.max()-x.min())
```

```python
df1 = df
```

```python
df1['temp'] = df1.groupby('Department')['Salary'].transform('mean')
```

```python
df1[['Department','temp']].head(10)
```

# Pivot table

```python
pd.pivot_table(df,
               'Salary',
               'Department',
               aggfunc='mean')
```

```python
pd.pivot_table(df,
               'Salary',
               index='Department',
               columns='Gender',
               aggfunc='mean')
```

```python
pd.pivot_table(df,
               values=['Salary','Age'],
               index='City',
               columns=['Gender'],
               aggfunc=['mean','min','max'],
               )
```

```python
pd.pivot_table(df,
               values=['Salary'],
               index='City',
               columns=['Gender'],
               aggfunc=['sum'],
               margins=True
               )
```

# Cross tab

```python
pd.crosstab(df['Gender'],df['Department'])
```

```python
pd.crosstab(df['Gender'],df['Department'],margins=True)
```

```python
pd.crosstab(df['City'],df['Department'],margins=True,normalize='columns')
```
