# Write the data

```python
sale = pd.read_excel("dataset.xlsx",sheet_name='Fact_Sales')
products = pd.read_excel("dataset.xlsx",sheet_name='Dim_Products')
customers = pd.read_excel("dataset.xlsx",sheet_name='Dim_Customers')

```

```python
sale.info()
```

```python
sale.to_csv("sale.csv",index=False)
```

```python
sale.head()
```

```python
sale.to_excel("sale.xlsx")
```

```python
sale.to_excel("cmpny.xlsx",sheet_name="Orders")
```

```python
products.to_excel("cmpny.xlsx",sheet_name="Products")
```

```python
with pd.ExcelWriter("company.xlsx") as writer:
    sale.to_excel(
        writer,
        sheet_name="Orders",
        index=False
    )
    products.to_excel(
        writer,
        sheet_name="Products",
        index=False
    )
    customers.to_excel(
        writer,
        sheet_name="Customers",
        index=False
    )
```
