# Datastructure in pandas

1. Series(1D)
2. Dataframe(2D)

```python
employees = {
    'emp_name':['rahul','neha','ram','sohit'],
    'salary':[32000,33000,34000,20000],
    'age':[23,22,24,25]
}
```

```python
df = pd.DataFrame(employees)
```

```python
print(df)
print(type(df))
```

```python
print(df['age'])
print(type(df['age']))
```

```python
salary = [34000,35000,23000,34000]

sr = pd.Series(salary)
print(sr)
print(type(sr))
```

### Series Functions

```python
slry = [34000,35000,23000,34000]
# salary = pd.Series(slry,index=[x for x in range(1,5)])
salary = pd.Series(slry,index=['a','b','c','d'])


print(salary)
```

```python
print(salary.dtypes)
```

```python
print("salary pool:",salary.sum())
```

```python
print("salary avg:",salary.mean())
```

```python
print("salary std:",salary.std())
```

```python
print("Descriptive information")

print(salary.describe())
```
