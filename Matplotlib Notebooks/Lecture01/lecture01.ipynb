```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
days = ["sun","mon","tue","wed","thu","fri","sat"]
sales = np.random.randint(20,50,7)
profit = np.random.randint(5,10,7)
```

```python
sales
```

# Line

```python
plt.figure(figsize=(10,5))

plt.plot(days,sales,
color='red',
marker='s',
linewidth=3,
markerfacecolor="blue",
markeredgecolor="green",
markersize=10,
label="Sale"
)
plt.plot(days,profit,
color='blue',
marker='^',
linewidth=3,
markerfacecolor="red",
markeredgecolor="green",
markersize=10,
label='profit'
)

plt.title("Sale & profitof days",fontsize=16)
plt.xlabel("Days")
plt.ylabel("Sale & profit in (thousand(₹))")
plt.grid()
plt.legend()
plt.show()
```

# Bar chart

```python
plt.figure(figsize=(10,5))

plt.bar(days,sales,width=0.5,edgecolor='black')
plt.title("Sale comparision of days")
plt.xlabel("days")
plt.ylabel("Sale in thousand (₹)")
plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.barh(days,sales,height=0.3,edgecolor='black')
plt.title("Sale comparision of days")
plt.xlabel("Sale in thousand (₹)")
plt.ylabel("days")
plt.grid(alpha=0.3)
plt.show()
```

# Grouped bar chart

```python
x = np.arange(len(days))

width = 0.4

plt.figure(figsize=(10,5))

plt.bar(x - width/2,sales,width,label='sales')
plt.bar(x + width/2,profit,width,label='profit')
plt.legend()
plt.title("Profit and Sales in days")
plt.xlabel("days")
plt.ylabel("profit and sales in thousand (₹)")
plt.show()
```

# Stacked bar chart

```python
remaining = sales - profit

plt.figure(figsize=(10,5))
plt.bar(days,profit,label='Profit')
plt.bar(days,remaining, bottom=profit,label='remaining sales')
plt.legend()
plt.show()
```

# Scatter chart

```python
sale = np.random.normal(150,10,500)
marketing = np.random.normal(50,20,500)
```

```python
plt.figure(figsize=(10,6))

sct = plt.scatter(sale,
marketing,
c=marketing,
alpha=0.5
)
plt.colorbar(sct)
plt.xlabel("sale")
plt.ylabel("marketing spending")
plt.title("marketing spending vs sale")
plt.show()
```

# Histogram plot

```python
scores = np.random.normal(100,10,1000)
```

```python
plt.figure(figsize=(10,5))
plt.hist(scores,bins=20,edgecolor='white',density=True)
plt.xlabel("distribution")
plt.ylabel("frequency")
plt.show()
```

# Pie chart

```python
category = ["Furniture","Toys","Accessories","Home Appliances","Office"]
quantity = [34,23,67,45,20]
```

```python
plt.figure(figsize=(7,7))

plt.pie(quantity,
labels=category,
autopct="%1.1f%%",
shadow=True,startangle=180,
explode=[0,0,0.1,0,0])

plt.show()
```
