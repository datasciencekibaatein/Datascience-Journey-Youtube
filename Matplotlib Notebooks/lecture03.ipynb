```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("food_delivery_orders.csv")
```

```python
df.head()
```

```python
df.dtypes
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
```

```python
df['month'] = df['order_date'].dt.month
```

```python
df.head()
```

```python
sns.__version__
```

```python
sns.color_palette("colorblind6")
```

```python
sns.set_theme(style="darkgrid",context="talk",palette="colorblind")
```

```python
plt.figure(figsize=(16,8))
sns.scatterplot(data=df,x='distance_km',y="delivery_time_min",hue="customer_type",size="distance_km")
plt.xlabel("distance in KM")
plt.ylabel("Delivery Time")
plt.title("Delivery VS Distance(KM)",fontsize="16",fontweight="bold")
plt.legend(loc="lower right")
plt.show()
```

# Line chart

```python
plt.figure(figsize=(16,6))
sns.lineplot(data=df,x='month',y='order_value',hue="city")
plt.xlabel("Months")
plt.ylabel("order value")
plt.title("Monthly order value")
plt.show()
```

# Bar plot

```python
plt.figure(figsize=(10,5))

sns.barplot(data=df,x='city',y="order_value",hue="payment_method")
plt.title("Order value by city")
plt.show()
```

# Count plot

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df,y='city',hue="customer_type")
plt.title("Frequency of orders by city")
plt.xlabel("frequency")
plt.ylabel("City")
plt.show()
```

```python
fig,axes = plt.subplots(figsize=(16,7),nrows=1,ncols=2)

sns.lineplot(data=df,x='month',y='order_value',ax=axes[0])
sns.barplot(data=df,x="payment_method",y='order_value',ax=axes[1])
plt.show()
```
