# Distribution

A distribution describes how the values of a variable are spread across different possible values or ranges.

In simple words, it tells us:
1. Which values occur frequently?
2. Which values occur rarely?
3. Where are most observations concentrated?
4. How spread out are the observations?
5. Is the data balanced or skewed?


# Distribution Shape

We generally discuss three basic shapes:
1. Symmetric distribution mean = mode = median
2. Right-skewed distribution (positive skew) mean > median > mode
3. Left-skewed distribution (negative skew) mean < median < mode


# Outliers
Outliers are data points that differ significantly from the rest of a dataset, lying an abnormal distance from other values and deviating from the general pattern.

# Histogram Plot

```python
df['order_value'].min()

mn = df['order_value'].mean()
md = df['order_value'].median()
mdd = df['order_value'].mode()[0]
```

```python
plt.figure(figsize=(10,5))
sns.histplot(data=df,x = "order_value",bins=20,kde=True)
plt.axvline(mn,linestyle="--",color="red",label=f"mean: {mn}")
plt.axvline(md,linestyle="--",color="green",label=f"median: {md}")
plt.axvline(mdd,linestyle="--",color="blue",label=f"mode: {mdd}")

plt.legend(loc="upper right")
plt.show()
```

# Box plot

```python
plt.figure(figsize=(10,7))
sns.boxplot(data=df,x ='city',y = "order_value",showmeans=True,width=0.5)
plt.show()
```
