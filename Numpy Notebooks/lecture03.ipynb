```python
matrix = np.array([[34,45,56],
                   [43,78,78],
                   [56,67,87]])

print(matrix * 2)
```

row broadcasting

```python
row = np.array([2,4,5])

print(matrix * row)
```

column broadcasting

```python
column = np.array([[2],
                   [3],
                   [4]])

print(matrix + column)
```

in row broadcast.. number of columns should be same

and in column broadcast.. number of rows should be same
