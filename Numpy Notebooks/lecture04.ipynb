# Probability Distribution

```python
np.random.normal(25,5,20)
```

### normal distribution

```python
import matplotlib.pyplot as plt


plt.hist(np.random.normal(25,5,1000),30)
plt.axvline(x=25,color='red')
plt.show()
```

### Uniform distribution

```python
import matplotlib.pyplot as plt


plt.hist(np.random.uniform(-10,10,1000),100)
plt.show()
```

## Dummy dataset

```python
n_students = 200

math_marks = np.random.randint(30,100,size=n_students)

physics_marks = math_marks + np.random.randint(-10,10,size=n_students)
physics_marks = np.clip(physics_marks,0,100)

english_marks = np.random.uniform(30,100,size=n_students)
english_marks = np.clip(english_marks,0,100)

print("Math: ",math_marks[:20])
print("Physics: ",physics_marks[:20])
print("English: ",english_marks[:20])

```
