```python
import numpy as np
```

```python
np.random.seed(42)
subjects = ["English","hindi","Math","Physics","Chemistry"]
students = [f"Student {i + 1}" for i in range(20)]
marks = np.random.randint(1,100,size=(20,5))
```

```python
marks
```

```python
print(f"Total Students: {marks.shape[0]}")
print(f"Total Subjects: {marks.shape[1]}")
```

total Marks of students

```python
total_marks = np.sum(marks,axis=1)
print(total_marks)
```

```python
percentages = total_marks/500 * 100
print(percentages)
```

```python
topper_idx = np.argmax(total_marks)
print(f"Topper Student{topper_idx + 1} with {total_marks[topper_idx]}" )
```

Top 3 students

```python
top_3_students = np.argsort(total_marks)[-3:][::-1]

for i,idx in enumerate(top_3_students,1):
    print(f"{i}. Student{idx} with {total_marks[idx]}")
```

Average subjects marks

```python
avg_marks = marks.mean(axis=0)

for subject,m in zip(subjects,avg_marks):
    print(f"{subject} : {m}")
```
