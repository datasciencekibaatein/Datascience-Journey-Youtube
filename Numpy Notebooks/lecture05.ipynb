Concatenate

```python
import numpy as np
```

```python
arr1 = np.array([2,3,4])
arr2 = np.array([3,4,6])


res = np.concatenate([arr1,arr2],axis=0)
print(res)
```

```python
mat1 = np.array([[2,3],
                 [3,4]])

mat2 = np.array([[5,2],
                 [4,8]])

res1 = np.concatenate([mat1,mat2],axis=1)

print(res1)
```
